### Step 4: Unique Features Inference Test (XAI, Ambiguity, Low Confidence & Code-Mixing)

```python
predictor = EduIntentPredictor(models_dir=os.path.join('..', 'models'))

test_queries = [
    'Why am I getting a KeyError while accessing a Pandas column?', # Standard Query
    'How can I implement CNN using Python?',                        # Ambiguous (DL + Python)
    'Can you explain this?',                                        # Low Confidence Warning
    'Pandas ma KeyError aave chhe su karvu?',                       # Gujarati Code-Mixed
    'What is deadlock in Operating Systems?'                        # Standard OS Query
]

for q in test_queries:
    res = predictor.predict(q)
    print(f"Query: '{res['query']}'")
    print(f"-> Language: {res['language']}")
    print(f"-> Prediction: {res['predicted_category']} (Confidence: {res['confidence']*100:.1f}%)")
    print(f"-> Key Terms (XAI): {res['important_terms']}")
    print(f"-> System Message: {res['user_message']}")
    print('-'*65)
```
